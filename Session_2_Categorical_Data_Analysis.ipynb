### Bar Chart

- under matplotlib
- plt.bar

```python
continent_keys=visa_df['continent'].value_counts().keys()
continent_values=visa_df['continent'].value_counts().values

cols=['yellow','blue','red','green','orange','pink']
plt.figure(figsize=(10,4))
plt.bar(continent_keys,
        continent_values,
       color=cols)
plt.xlabel('Continent')
plt.ylabel('No of Applicants')
plt.title('Continent_ bar')
```

```python
continent_keys=visa_df['continent'].value_counts().keys()
continent_values=visa_df['continent'].value_counts().values

cols=['yellow','blue','red','green','orange','pink']
plt.figure(figsize=(10,4))
plt.bar(continent_keys,
        continent_values,
       color=cols)
plt.xlabel('Continent')
plt.ylabel('No of Applicants')
plt.title('Continent_ bar')
plt.savefig('Continent_bar.jpg')   # filename.extension
plt.show()  # remove the text
```
